```python
# Student grade system: A = 90 and above, B = 80 and above, C = 70 and above, D = 60 and above, F = below 60
# if elif else format
marks = int(input("Enter your marks: "))
if marks >= 90:
    print("Grade = 'A'")
elif marks >= 80:
    print("Grade = 'B'")
elif marks >= 70:
    print("Grade = 'C'")
elif marks >= 60:
    print("Grade = 'D'")
else:
    print("Grade = 'F'")
```
